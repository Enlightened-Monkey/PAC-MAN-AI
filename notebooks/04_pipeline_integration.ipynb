# 04 – Pipeline Integration

Combine the pre-trained CNN detector with the trained RL agent for end-to-end inference.

```python
import sys
sys.path.insert(0, '..')
import numpy as np
from src.environment.pacman_env import PacmanEnv, _OBS_SIZE
from src.models.cnn_detector import ObjectDetector
from src.models.rl_agent import RLAgent
```

```python
env = PacmanEnv(seed=0)
detector = ObjectDetector(obs_size=_OBS_SIZE)
# agent = RLAgent.load('../models/dqn_pacman', env=env, algorithm='dqn')
```

```python
# End-to-end inference loop (Stage 2)
# 1. Capture a game frame (simulated here with a random array)
# 2. CNN maps frame -> state vector
# 3. RL agent selects action from state vector

for step in range(5):
    fake_frame = np.random.randint(0, 255, (84, 84, 3), dtype=np.uint8)
    state_vector = detector.predict(fake_frame)
    # action, _ = agent.predict(state_vector)
    action = env.action_space.sample()  # placeholder until agent is trained
    obs, reward, terminated, truncated, info = env.step(action)
    print(f'Step {step}: action={action}, reward={reward:.1f}, score={info["score"]}')
    if terminated or truncated:
        break
```
